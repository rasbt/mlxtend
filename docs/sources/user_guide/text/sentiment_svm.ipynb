# SentimentSVM

`SentimentSVM` is a text-classification estimator that combines
TF-IDF feature extraction with a linear Support Vector Machine.

It can classify text into two or more categories, including
positive, negative, and neutral sentiment.

## How it works

Text data must be converted into numerical features before a machine
learning model can use it.

`SentimentSVM` performs two steps:

1. `TfidfVectorizer` converts text into numerical TF-IDF features.
2. `SVC` with a linear kernel learns to classify the text.

The two steps are combined in a scikit-learn pipeline.

```python
import sys
from pathlib import Path

repo_root = Path.cwd()

while repo_root.name != "mlxtend" and repo_root != repo_root.parent:
    repo_root = repo_root.parent

sys.path.insert(0, str(repo_root))

from mlxtend.text import SentimentSVM

print(SentimentSVM)
```

```python
training_text = [
    "This product is excellent",
    "I really love this product",
    "The experience was wonderful",
    "This product is terrible",
    "I hate this product",
    "The experience was awful",
    "The package arrived today",
    "The product is available",
    "The item is standard",
]

training_labels = [
    "positive",
    "positive",
    "positive",
    "negative",
    "negative",
    "negative",
    "neutral",
    "neutral",
    "neutral",
]
```

## Train the classifier

The classifier learns the relationship between the text documents
and their sentiment labels using the `fit` method.

```python
classifier = SentimentSVM()
classifier.fit(training_text, training_labels)
```

## Make predictions

After training, the classifier can predict the sentiment of new text.

```python
new_text = [
    "I enjoyed this excellent product",
    "This was a terrible experience",
    "The item arrived today",
]

predictions = classifier.predict(new_text)
predictions
```

## Evaluate the classifier

The `score` method returns the mean accuracy on labeled data.
This example evaluates the classifier on the small demonstration
dataset used for training.

```python
accuracy = classifier.score(training_text, training_labels)
print(f"Training accuracy: {accuracy:.2f}")
```

## Important considerations

This example uses a small demonstration dataset only to explain the
API. A real sentiment-analysis application should use a larger,
representative dataset and separate training and test data.

The `C` parameter controls SVM regularization. The `ngram_range`
parameter can be changed to include word pairs or longer phrases.
